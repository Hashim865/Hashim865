### Cleaning the `rate` column

The `rate` column is currently stored as an object, but it contains numerical ratings. We need to convert it to a numeric type. We will extract the numerical part and handle non-numeric entries.

```python
import numpy as np

def clean_rate(rate):
    if isinstance(rate, str):
        if ' ' in rate:
            # Handle cases like '4.1 /5'
            rate = rate.split(' ')[0]
        if '/' in rate:
            return float(rate.split('/')[0])
        elif rate == 'NEW' or rate == '-' or rate == 'nan':
            return np.nan
        else:
            try:
                return float(rate)
            except ValueError:
                return np.nan
    return rate


df['rate'] = df['rate'].apply(clean_rate)

print('Data types after cleaning the rate column:')
display(df.info())
print('\nMissing values in rate column after cleaning:')
display(df['rate'].isnull().sum())
print('\nFirst 5 rows of the cleaned rate column:')
display(df['rate'].head())
```
